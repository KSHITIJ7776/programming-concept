```python
class Student:
    def __init__(self, name,age,course):
        self.name = name
        self.age = age
        self.course = course

    def display(self):
        print("Student Name:", self.name)
        print("Student age:", self.age)
        print("Student course:", self.course)

# Creating an object
s1 = Student("Krunal",22,"itiss")

# Calling method
s1.display()
```
