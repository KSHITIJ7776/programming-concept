```python
# Function to calculate factorial using recursion

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

# Example
num = 5
print("Factorial =", factorial(num))
```

```python
# Function to check whether a number is prime

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

# Example
num = 17

if is_prime(num):
    print(num, "is a Prime Number")
else:
    print(num, "is not a Prime Number")
```

```python
# Function to find the sum of even numbers in a list

def sum_even(numbers):
    total = 0

    for num in numbers:
        if num % 2 == 0:
            total += num

    return total

# Example
nums = [10, 15, 20, 25, 30]

print("Sum of even numbers:", sum_even(nums))
```

```python
# Function to count vowels and consonants

def count_vowels_consonants(text):
    vowels = "aeiouAEIOU"
    v_count = 0
    c_count = 0

    for ch in text:
        if ch.isalpha():  # Check if character is a letter
            if ch in vowels:
                v_count += 1
            else:
                c_count += 1

    return v_count, c_count

# Example
string = "Hello World"

vowels, consonants = count_vowels_consonants(string)

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
# Function to calculate x^y without using built-in power functions

def power(x, y):
    result = 1

    for i in range(y):
        result *= x

    return result

# Example
x = 2
y = 5

print("Result:", power(x, y))
```
