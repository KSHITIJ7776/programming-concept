```python
products = {}

products["Laptop"] = 10
products["Mouse"] = 3
products["Keyboard"] = 2

products["Mouse"] = 5

print("Low Stock Products:")
for product, qty in products.items():
    if qty < 5:
        print(product, ":", qty)
```

```python
students = {}

# Add students
students[101] = "Rahul"
students[102] = "Priya"
students[103] = "Amit"

# Update student
students[102] = "Neha"

# Delete student
del students[103]

# Display all records
print("Student Records:")
for id, name in students.items():
    print(id, ":", name)
```

```python
# Marks of 10 students
marks = [78, 85, 92, 67, 88, 74, 95, 81, 69, 90]

# Highest mark
print("Highest Mark:", max(marks))

# Lowest mark
print("Lowest Mark:", min(marks))

# Average mark
average = sum(marks) / len(marks)
print("Average Mark:", average)
```

```python
# Shopping Cart Management

cart = []

# Add items
cart.append("Milk")
cart.append("Bread")
cart.append("Eggs")

# Remove an item
cart.remove("Bread")

# Display all items
print("Shopping Cart:")
for item in cart:
    print(item)
```

```python
# Employee Records

emp1 = (101, "Rahul", "HR", 40000)
emp2 = (102, "Priya", "IT", 55000)
emp3 = (103, "Amit", "Finance", 50000)

employees = [emp1, emp2, emp3]

# Display all employee details
print("Employee Details:")
for emp in employees:
    print(emp)

# Find employee with highest salary
highest = max(employees, key=lambda x: x[3])

print("\nEmployee with Highest Salary:")
print(highest)
```

```python
# Coordinate Points

points = [(3, 4), (1, 2), (5, 6), (2, 1)]

# Display all points
print("Coordinate Points:")
for p in points:
    print(p)

# Find point nearest to origin (0,0)
nearest = min(points, key=lambda p: p[0]**2 + p[1]**2)

print("\nNearest Point to Origin:")
print(nearest)
```
