```python
# List of 10 numbers
numbers = [12, 45, 7, 89, 23, 56, 34, 90, 11, 67]

# Find largest and smallest numbers
largest = max(numbers)
smallest = min(numbers)

print("Largest Number:", largest)
print("Smallest Number:", smallest)
```

```python
# List with duplicate elements
numbers = [10, 20, 10, 30, 20, 40, 50, 40]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("List after removing duplicates:")
print(unique)
```

```python
# Original list
numbers = [10, 20, 30, 40, 50]

# Reverse the list without reverse()
reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed List:")
print(reversed_list)
```

```python
# Two lists
list1 = [30, 10, 50]
list2 = [20, 40, 60]

# Merge lists
merged_list = list1 + list2

# Sort merged list
merged_list.sort()

print("Merged and Sorted List:")
print(merged_list)
```
